### Safely Handling Missing Data

```python
def handlingMissingValues(df):
    df["base price (in lacs)"] = df["base price (in lacs)"].fillna(df["base price (in lacs)"].mean())
    df["final price (in lacs)"] = df["final price (in lacs)"].fillna(0)
    df["franchise"] = df["franchise"].fillna("Unsold")
    return df
```

### Feature Engineering


Feature engineering is the process of transforming raw data into relevant features that help improve the performance of machine learning models. It involves selecting, creating, and adjusting data attributes to make the model's predictions more accurate.
 In the context of Exploratory Data Analysis (EDA), feature engineering plays a crucial role as it uses insights gained from EDA to create better representations of the data, allowing models to learn more effectively from the data.
 This process can include techniques such as handling missing values, creating interaction terms, and re-scaling numeric features, all of which are often inspired by discoveries made during the EDA phase

```python
df["Demand Index"] = df["final price (in lacs)"]/df["base price (in lacs)"]
```
