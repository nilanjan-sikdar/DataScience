```python
room_type_counts = df['room_type'].value_counts()


plt.bar(room_type_counts.index, room_type_counts.values, color='skyblue', edgecolor='black')


plt.xlabel('Room Type')
plt.ylabel('Count')
plt.title('Room Type Distribution')
plt.xticks(rotation=45)  
plt.tight_layout()
plt.show()
```

```python

room_type_counts = df['room_type'].value_counts()


plt.pie(
    room_type_counts.values,             # values
    labels=room_type_counts.index,       # labels
    autopct='%1.1f%%',                   # show percentages
    startangle=140,                      # rotate start angle
    shadow=True,                         # shadow effect
    explode=(0.1,) + (0,) * (len(room_type_counts)-1)  # explode 1st slice
)

plt.title('Room Type Distribution')
plt.axis('equal')  
plt.show()
```
