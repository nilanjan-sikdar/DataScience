# Mathplotlib

```python
from matplotlib import pyplot as plt
import pandas as pd
```

```python
sample_data = {
    'quarter' : ['Q1 2022', 'Q2 2022', 'Q3 2022', 'Q1 2023', 'Q2 2023'],
    'Fridge' : [12.5, 11, 13.2, 16.4, 13.8],
    'Dishwasher' : [8.0, 7.5, 9.0, 11.0, 9.5],
    'Washing Machine' : [9.2, 8.5, 7.5, 7.6, 6.8],
}
```

```python
df = pd.DataFrame(sample_data)
```

```python
df.to_csv("sample_data.csv")
```

## plot Chart

```python
# size of the plots in (x , y)
plt.figure(figsize=(20,6))
# plots the value(xasix, yaxis, color_lines, label_lines_identify) 
plt.plot(df['quarter'], df['Fridge'],color= '#ff9933', label = 'Fridge')
plt.plot(df['quarter'], df['Dishwasher'],color= 'grey', label = 'Dishwasher')
plt.plot(df['quarter'], df['Washing Machine'], color= 'green', label = 'Washing Machine')
# title of the plot
plt.title('product Sale')
# x and y label of plot
plt.ylabel('Revenue in $')
plt.xlabel('yearwise')
# make a table of data look top right below plot
plt.legend()
# to show the plot
plt.show()
```

## pie Chart

```python
total_revenue = df[['Fridge', 'Dishwasher', 'Washing Machine']].sum()
total_revenue
total_revenue.index
```

```python
plt.pie(total_revenue, labels = total_revenue.index, autopct='%1.2f%%', explode=(0.2, 0.1,0), shadow=True, startangle=90)
plt.show()
```

## bar Chart

```python
df.plot(kind='bar', x='quarter')
plt.xticks(rotation=45)
plt.show()
```

## another definition of bar Chart

```python
df2 = df.set_index('quarter')
df2.plot(kind='bar')
plt.xticks(rotation = 45)
plt.show()
```

## histogram Charts

```python
df2.plot(kind='barh')
plt.yticks(rotation=45)
plt.show()
```
